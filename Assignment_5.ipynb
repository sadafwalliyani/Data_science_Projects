```python
import numpy as np

# Step 1: Load and Explore the Dataset
students = np.array(["Alice", "Bob", "Charlie", "David", "Emma"])
scores = np.array([
    [85, 78, 92],
    [88, np.nan, 80],
    [75, 85, 79],
    [np.nan, 90, 82],
    [89, 76, 91]
])
attendance = np.array([90, 85, 95, np.nan, 88])

```

```python
# Print shape, size, and data type
print("Shape of scores:", scores.shape)
print("Size of scores:", scores.size)
print("Data type of scores:", scores.dtype)

```

```python
# Step 2: Data Cleaning & Handling Missing Values
nan_mask_scores = np.isnan(scores)
nan_mask_attendance = np.isnan(attendance)
```

```python


# Replace missing values in scores with the average of each column
col_means = np.nanmean(scores, axis=0)
scores[nan_mask_scores] = np.take(col_means, np.where(nan_mask_scores)[1])
```

```python
# Replace missing values in attendance with mean attendance
attendance[nan_mask_attendance] = np.nanmean(attendance)
```

```python
# Step 3: Perform Mathematical Analysis
mean_scores = np.mean(scores, axis=0)
median_scores = np.median(scores, axis=0)
std_scores = np.std(scores, axis=0)
max_scores = np.max(scores, axis=0)
average_student_scores = np.mean(scores, axis=1)
```

```python
# Identify top performer
top_performer_index = np.argmax(np.sum(scores, axis=1))
top_performer = students[top_performer_index]
```

```python
# Step 4: Reshaping & Broadcasting
# Normalize scores using min-max scaling
min_scores = np.min(scores, axis=0)
max_scores = np.max(scores, axis=0)
norm_scores = (scores - min_scores) / (max_scores - min_scores)
```

```python
# Broadcast bonus 5 marks to students scoring below 80 in any subject
scores = np.where(scores < 80, scores + 5, scores)
```

```python
# Step 5: Stacking & Splitting Data
stacked_data = np.hstack((scores, attendance.reshape(-1, 1)))
split_data = np.array_split(stacked_data, 2)

```

```python
# Step 6: Filtering Data with Boolean Conditions
low_attendance_students = students[attendance < 90]
high_scorers = students[np.all(scores > 85, axis=1)]
failed_students = students[np.any(scores < 40, axis=1)]

```

```python
# Print Results
print("\nCleaned Scores:\n", scores)
print("\nMean Scores:", mean_scores)
print("Median Scores:", median_scores)
print("Standard Deviation:", std_scores)
print("Max Scores:", max_scores)
print("Top Performer:", top_performer)
print("\nStudents with Low Attendance:", low_attendance_students)
print("Students Scoring Above 85 in All Subjects:", high_scorers)
print("Students Who Failed:", failed_students)
```
