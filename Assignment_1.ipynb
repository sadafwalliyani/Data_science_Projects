`Part 1: Temperature Comparison`
Problem Statement:
Write a program that:
● Takes input for the temperatures of two cities in Celsius.
● Compares the temperatures using relational operators (>, <, ==, !=).
Prints a message like:
City A is hotter than City B.

```python
input = __builtins__.input
temp_city_a = float(input('What is the temperature in City A (Celsius)? ').replace('°C', ''))
temp_city_b = float(input('What is the temperature in City B (Celsius)? ').replace('°C', ''))


# Compare the temperatures
if temp_city_a > temp_city_b:
    print('City A is hotter than City B.')
elif temp_city_a < temp_city_b:
    print('City B is hotter than City A.')
else:
    print('City A and City B have the same temperature.')
```

###Part 2: Bill Splitter
Problem Statement:
Write a program that calculates how much each person needs to pay when splitting a bill:
1. Take the total bill amount as input.
2. Take the number of people as input.
3. Calculate each person’s share by dividing the total amount by the number of people.
Print the result in this format:
Total Bill: $[amount]
Number of People: [people]
Each Person Pays: $[share]

```python
input = __builtins__.input
total_amount= int(input('Whats the total amount:'))
num_people = int(input('How many person are there? '))
total_amount_per_person = total_amount / num_people
print(f'Total Bill: RS {total_amount}')
print(f'Number of People: {num_people}')
print(f'Each Person Pays: RS {total_amount_per_person}')
```

###Part 3: Custom Message Formatter
`Problem Statement:`
Write a program that takes the following inputs:

● A name

● A favorite color

● A favorite number

[Number].
