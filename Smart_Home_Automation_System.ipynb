### 📌 Scenario:
You are developing a Smart Home Automation System that allows users to control and
monitor various home appliances using Python data structures like Lists, Tuples,
Dictionaries, Sets, and Functions.
Your task is to implement different functionalities step by step.

### 1. Device Management (Lists & Tuples)
`Task 1.1:`
Create and Manage a List of Smart Devices
● Create a list of at least 5 smart home devices (e.g., "Lights", "Fan", "AC", "Door Lock", "Thermostat").

● Allow the user to:
1.   Add a new device.
2.   Remove a device
3.  Sort devices alphabetically


```python
# Task 1.1: Create and Manage a List of Smart Devices
smart_devices = ["Lights", "Fan", "AC", "Door Lock", "Thermostat"]
print(f"Devices: {smart_devices}")
```

```python
# Add a new device
smart_devices.append("Security Camera")
print(f"Updated Devices: {smart_devices}")
```

```python
# Sort devices alphabetically
sorted_devices = sorted(smart_devices)
print(f"Sorted Devices: {sorted_devices}\n")
```

```python
# Task 1.2: Store Device Settings Using Tuples
device_settings = [
    ("Lights", "ON", 60),
    ("Fan", "OFF", 40),
    ("AC", "ON", 1500),
    ("Door Lock", "OFF", 5),
    ("Thermostat", "ON", 200),
    ("Security Camera", "ON", 10)
]
print("Device Status and Power Consumption:")
for device in device_settings:
    print(f"{device[0]}: {device}")
print()
```

```python
# Task 2.1: Store Device Energy Usage in a Dictionary
energy_usage = {"Lights": 5, "AC": 120, "Fan": 30}
print(f"Initial Energy Usage: {energy_usage}")
```

```python
# Update existing usage and add new device
energy_usage["AC"] = 110
energy_usage["Thermostat"] = 20
print(f"Updated Energy Usage: {energy_usage}")
print(f"Total Energy Consumption: {sum(energy_usage.values())} kWh\n")
```

```python
# Task 2.2: Unique Power-Saving Modes Using Sets
power_modes = {"Eco Mode", "Night Mode"}
print(f"Power-Saving Modes: {power_modes}")
```

```python
# Add new mode and display
power_modes.add("Away Mode")
print(f"New Mode Added: 'Away Mode'")
print(f"Updated Modes: {power_modes}\n")
```

```python
# Task 3.1: Calculate Monthly Energy Cost
def calculate_energy_cost(energy_usage, rate_per_kwh):
    total_cost = energy_usage * rate_per_kwh
    return f"Total Monthly Energy Cost: ${total_cost:.2f}"

print(calculate_energy_cost(165, 0.12))
print()
```

```python
# Task 3.2: Find Common Devices in Two Homes
def find_common_devices(home1, home2):
    return set(home1) & set(home2)

home1_devices = ["Lights", "AC", "Fan"]
home2_devices = ["Fan", "Door Lock", "Lights"]
print(f"Common Devices: {find_common_devices(home1_devices, home2_devices)}\n")
```

```python
# Task 4.1: Create Automation Rule
def set_automation_rule(device, time, action):
    return f"Automation Rule: {device} will be turned {action} at {time}."

print(set_automation_rule("Lights", "10:00 PM", "OFF"))
print()
```

```python
# Task 4.2: Optimize Power Consumption
def optimize_power(devices):
    return [device[0] for device in devices if device[1] <= 1000]

sample_devices = [("Lights", 60), ("AC", 1500), ("Fan", 40), ("Heater", 2000)]
print(f"Active Devices: {optimize_power(sample_devices)}\n")
```

```python
# Bonus Challenge: Smart Home AI Assistant
def smart_home_assistant(name, *args, **kwargs):
    print(f"Checking status for: {', '.join(args)}")
    kwargs_lower = {k.lower(): v for k, v in kwargs.items()}
    for device in args:
        lower_device = device.lower()
        if lower_device in kwargs_lower:
            print(f"{device} set to {kwargs_lower[lower_device].upper()}")

smart_home_assistant("Alice", "Lights", "AC", lights="OFF", ac="ON")
```
