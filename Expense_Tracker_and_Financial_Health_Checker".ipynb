### Objective:
The goal of this project is to build a simple Expense Tracker that allows a user to input their
income, expenses, and savings goals. The program will then provide real-time insights into
their financial health by analyzing their data using strings, operators, and conditional
statements (including nested if-else).

## Project Requirements:
1. The program should be interactive and take inputs from the user.
2. It must validate the inputs (e.g., ensure numerical values for income and expenses).
3. Provide meaningful output based on the user’s input and decisions.
4. Demonstrate usage of strings, operators, and if-else (with nested conditions).

### Project Features:
1. User Profile Setup:
● Ask the user to input their name and profession.
● Display a personalized welcome message.

```python
user_name = input("Write your name: ")
user_proession= input("Write your proeession: ")
print(f"Welcome {user_name} ! Let's analyze your financial health as a {user_proession}")

"""Income and Expense Management:
● Ask the user for their monthly income and monthly expenses.
● Calculate the savings (Income - Expenses).
● Use arithmetic operators to calculate the percentage of income saved."""

monthly_income = float(input("Enter your monthly income: "))
monthly_expenses = float(input("Enter your monthly expenses: "))

savings = monthly_income - monthly_expenses
percentage_saved = (savings / monthly_income) * 100
"""
Financial Health Check:
● Use nested if-else conditions to provide insights based on the savings percentage:
○ If savings are greater than or equal to 20% of income:"""
if percentage_saved >= 20:
    print(f"Great job, {user_name}! You have a strong savings habit.")
elif 10 <= percentage_saved < 20:
    print(f"Good, {user_name}, but you could save a bit more.")
else:
    print(f"Warning, {user_name}: Your savings are too low. Consider cutting expenses!")

"""Categorize Expenses:
● Ask the user to categorize their expenses into three major areas:
1. Essentials (e.g., rent, utilities, groceries)
2. Wants (e.g., dining out, entertainment)
3. Savings/Investments
● Use string operations to display a summary of these categories."""

user_essentials= float(input('How much do you spend on Essentials?'))
user_spend= float(input('How much do you spend on Wants?'))
user_saving= float(input('How much do you spend on Savings/Investments?'))

if user_essentials >= 50:
    print(f"Great job, {user_name}! You have a strong savings habit.")
elif 10 <= user_essentials < 20:
    print(f"Good, {user_name}, but you could save a bit more.")
else:
    print(f"Warning, {user_name}: Your savings are too low. Consider cutting expenses!")

"""Custom Goals:
● Allow the user to input a savings goal (in percentage).
● Use conditional logic to evaluate if they are meeting their goal:
○ If the current savings percentage is greater than or equal to the goal:"""

savings_goal = float(input("Enter your savings goal (in %): ").strip('%')) / 100
if percentage_saved >= savings_goal:
  print(f"Congratulations, {user_name}! You've achieved your savings goal.")
else:
  print(f"Keep working on your savings, {user_name}. You're {percentage_saved - savings_goal}% away from your goal.")

""" Export Summary (Bonus Feature):
● Generate a summary of the user’s financial health in a formatted string or save it to a text file"""
summary = f"""Financial Health Summary for {user_name}:
Income: {monthly_income}
Expenses: {monthly_expenses}
Savings: {savings} ({percentage_saved:.2f}%)
Expense Breakdown:
Essentials: {user_essentials}
Wants: {user_spend}
Savings/Investments: {user_saving}
Savings Goal: {savings_goal}%
Status: {percentage_saved - savings_goal}% away from your goal."""

print(summary)
```
